```python
import argilla as rg
import os

HF_TOKEN = os.getenv("HF_TOKEN")  # only for private spaces

client = rg.Argilla(
    api_url="http://localhost:6900",
    api_key="argilla.apikey",
    headers={"Authorization": f"Bearer {HF_TOKEN}"},  # only for private spaces
)
```

```python
dataset = client.datasets(name="ag_news")
```

```python
status_filter = rg.Query(filter=rg.Filter([("status", "==", "completed")]))

filtered_records = dataset.records(status_filter)
filtered_records
```

```python
filtered_records.to_datasets().push_to_hub("goosmanlei/ag_news_annotated")
```

```python
dataset.to_hub(repo_id="goosmanlei/ag_news_annotated")
```
