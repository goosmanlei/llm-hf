试试看！ 使用像 mc4 或者 oscar 这样的大型 Common Crawl 语料库来创建一个流式多语言数据集，该数据集代表你选择的国家/地区语言的口语比例。例如，瑞士的四种民族语言分别是德语、法语、意大利语和罗曼什语，因此你可以尝试根据根据口语比例对 Oscar 子集进行抽样来创建一个瑞士语料库。

```python
from datasets import load_dataset, interleave_datasets

# 以瑞士为例，四种语言及大致口语比例：德语63%、法语23%、意大利语8%、罗曼什语<1%
# 用 probabilities 参数按比例抽样
lang_configs = ["de", "fr", "it"]
probabilities = [0.63, 0.23, 0.14]

datasets_list = [
    load_dataset("allenai/c4", lang, split="train", streaming=True)
    for lang in lang_configs
]
```

```python
next(iter(datasets_list[0]))
```

```python

swiss_dataset = interleave_datasets(datasets_list, probabilities=probabilities, seed=42)
```

```python

# 查看前3条
from itertools import islice
samples = list(islice(swiss_dataset, 3))
for s in samples:
    print(s["text"][:200])
    print("---")
```
